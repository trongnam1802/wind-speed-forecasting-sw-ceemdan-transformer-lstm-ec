```python
import pandas as pd
import numpy as np
from PyEMD import CEEMDAN
from tqdm import tqdm

# Định nghĩa đường dẫn tệp đầu vào và đầu ra
input_path = 
output_path =

# Đọc dữ liệu từ CSV
df = pd.read_csv(input_path)
speed = df["speed"].values  # Lấy giá trị từ cột "speed"

lookback = 160
num_windows = len(speed) - lookback + 1

window_imf_list = []

for i in tqdm(range(num_windows), desc="Online-style CEEMDAN (no stride)"):
    segment = speed[i : i + lookback]

    # Luôn chạy CEEMDAN mới tại mỗi bước
    ceemdan = CEEMDAN()
    imfs = ceemdan.ceemdan(segment)
    residual = ceemdan.residue

    # Lấy giá trị cuối cùng của mỗi IMF
    imf_last = imfs[:, -1].reshape(1, -1)
    imf_df = pd.DataFrame(imf_last, columns=[f"IMF_{j+1}" for j in range(imf_last.shape[1])])
    imf_df["residual"] = residual[-1] if isinstance(residual, (list, np.ndarray)) else residual
    imf_df["window_index"] = i

    window_imf_list.append(imf_df)

# Gộp kết quả
result_df = pd.concat(window_imf_list, ignore_index=True)
result_df.to_csv(output_path, index=False, encoding="utf-8-sig")

print(f"✅ Đã lưu xong {num_windows} cửa sổ CEEMDAN vào '{output_path}'")
```
